```python
import pandas as pd
import numpy as np
#the main problems with this datasets are the columns Quantity, UnitPrice, and TotalAmount
#Quantity has negative values,
#UnitPrice have prices that are greater than the TotalAmount and others that dont coordinate with the totalAmount and Quantity values
#TotalAmount has values that that don't coordinate with the Price and Quantity values
```

```python
# removing all negative values from the Quantity column and putting it back into the dataset
data = pd.read_csv('ecommerce_data.csv')
data.loc[:,'Quantity']=data.loc[:,'Quantity'].abs() 
```

```python
# With the new quantity values we can calculate a newUnitPrice column by dividing the TotalAmount by the Quality
data['newUnitPrice']=round(data['TotalAmount']/data['Quantity'],2)
data.head(10)
```

```python
#Checking to see which UnitPrice are different from the newUnitPrices
data[["TotalAmount", "UnitPrice", "Quantity","newUnitPrice" ]][data["UnitPrice"]!=round(data["newUnitPrice"],2)]
```

```python
#replacing UnitPrice with the new UnitPrice column to make the data more consistent, and removing the newUnitPrice column
data['UnitPrice']=data['newUnitPrice']
data=data.drop(columns='newUnitPrice')
```
